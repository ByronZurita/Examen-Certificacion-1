## Descripción del Dataset


El **Dataset de Salarios en Trabajos de Ciencia de Datos** contiene 11 columnas que proporcionan información detallada sobre los salarios en el campo de la ciencia de datos. A continuación se describen cada una de las columnas:

- **work_year**: El año en que se pagó el salario.
- **experience_level**: El nivel de experiencia en el trabajo durante el año.
- **employment_type**: El tipo de empleo para el rol desempeñado (tiempo completo, medio tiempo, contratista, etc.).
- **job_title**: El puesto de trabajo desempeñado durante el año.
- **salary**: El monto total bruto del salario pagado.
- **salary_currency**: La moneda en la que se pagó el salario, utilizando un código de moneda ISO 4217.
- **salary_in_usd**: El salario convertido a dólares estadounidenses (USD).
- **employee_residence**: El país de residencia principal del empleado durante el año laboral, según el código de país ISO 3166.
- **remote_ratio**: El porcentaje de trabajo realizado de forma remota.
- **company_location**: El país donde se encuentra la oficina principal o sucursal contratante del empleador.
- **company_size**: El número mediano de empleados que trabajaron en la empresa durante el año.

# Deteccion de datos anomalos

```python
plt.boxplot(df['salary_in_usd'], vert=False)
plt.title('Boxplot de Salarios en USD')
plt.xlabel('Salario')
plt.show()
```

```python
# Citado
# Usando la grafica anterior queremos calcular el primer quartil (q1) y el tercer quartil (3)
q1 = df['salary_in_usd'].quantile(0.25) # Quartil 1: 25% de los salarios son menores o iguales a este valor
q3 = df['salary_in_usd'].quantile(0.75) # Quartil 3: 75% de los salarion son menores o iguales a este valor
iqr = q3 - q1 # Rango inter: Diferencia entre q3 y q1
print('25% de los salarios son menores o iguales a:', q1)
print('75% de los salarios son menores o iguales a:', q3)
print('Rango inter:', iqr)
bi = q1 - 1.5 * iqr # Bigote inferor (Izquierda)
bs = q3 + 1.5 * iqr # Bigote superior (Derecha)
print('\nBigote inferior:', bi)
print('Bigote superior:', bs)
outliers = df[(df['salary_in_usd'] < bi) | (df['salary_in_usd'] > bs)]
print('Salarios que no esten dentro de los bigotes son anomalos')
print('\nNumero de salarios anomalos:\n',len(outliers))
outliers
```

```python
# Seleccionamos solo los outliers con un nivel de experencia no adecuada
junior_outliers = outliers[outliers['experience_level'] == 'Junior']
junior_outliers
```

```python
# Eliminamos los outliers seleccionados
df = df.copy()
df = df.drop(junior_outliers.index)
df
```

```python
# Restoracion del index
df = df.reset_index(drop=True)
df
```

# Visualizaciones exploratorias multivariadas

```python
# Muestra la relacion entre el raio de salarios y el ratio de trabajo remoto, sugiere que empleados con salarios mas altos tienen mas flexibilidad para trabajar de manera remota.
plt.scatter(df['salary_in_usd'], df['remote_ratio'])
plt.title('Scatterplot de Salario vs Remote Ratio')
plt.xlabel('Salario')
plt.ylabel('Remote Ratio')
plt.show()
```

# Identificacion de tendencias

```python
# La linea en medio de la caja representa la mediana de los salarios en USD, demostrando unos outliers en los salarios.
plt.boxplot(df['salary_in_usd'], vert=False)
plt.title('Boxplot de Salarios en USD')
plt.xlabel('Salario')
plt.show()
# Los puntos fuera de los bigotes son considerados outliers, estos significan salarios altos cuales podrian ser datos erroneos.
# Este mismo grafico podria ser comparado con datos de años anteriores, esto reflejaria la demanda en aumento del talento en el area tecnologica.
```

```python
# Similares cantidades de empleados trabajan de manera remota a los que trabajan presencial, demostrando la flexibilidad laboral.
plt.hist(df['remote_ratio'])
plt.title('Histograma de Remote Ratio')
plt.xlabel('Remote Ratio')
plt.ylabel('Frecuencia')
plt.show()
# La existencia de valores en 50% indica una parte de empleados tienen opciones flexibles cuales pueden atraer mas talentos en campos tecnologicos donde el talento es escaso.
```

```python
# Muestra la relacion entre el raio de salarios y el ratio de trabajo remoto, sugiere que empleados con salarios mas altos tienen mas flexibilidad para trabajar de manera remota.
plt.scatter(df['salary_in_usd'], df['remote_ratio'])
plt.title('Scatterplot de Salario vs Remote Ratio')
plt.xlabel('Salario')
plt.ylabel('Remote Ratio')
plt.show()
# La existencia de salarios altos en posiciones remotas puede atraer mas profesionales lo cual aumentaria la competicion por atraer estos talentos con mas compensaciones.
```

```python
# Demuestra que la experencia influye en la compensacion.
plt.scatter(df['experience_level'], df['salary_in_usd'])
plt.title('Scatterplot de Nivel de Experiencia vs Salario')
plt.xlabel('Nivel de Experiencia')
plt.ylabel('Salario')
plt.show()
# Demuestra una tendencia clara del incremento salarial con niveles de experencia, cuales sugiere que la acumulacion de experencia se traduce en mejores oportunidades salariales.
```
